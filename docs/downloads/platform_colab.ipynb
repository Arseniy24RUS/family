```python
from google.colab import files
uploaded = files.upload()
```

```python
import json, pathlib
from lab_reproduce import repeat
inputs = [p for p in pathlib.Path(".").glob("*.json") if p.name != "repeated_result.json"]
if len(inputs) != 1:
    raise ValueError("Загрузите ровно один JSON-экспорт лаборатории")
document = json.loads(inputs[0].read_text(encoding="utf-8"))
result = repeat(document)
print(json.dumps(result, ensure_ascii=False, indent=2))
```

```python
output = {"config": document.get("config", {}), "repeated_result": result, "reference_result": document.get("result")}
pathlib.Path("repeated_result.json").write_text(json.dumps(output, ensure_ascii=False, indent=2), encoding="utf-8")
files.download("repeated_result.json")
```
